```python
import os
import re
from datasets import load_dataset
from itertools import islice
from tqdm import tqdm

# Create output directory
os.makedirs("fineweb_data", exist_ok=True)

# English is in FineWeb v1 (not fineweb-2)
dataset = load_dataset("HuggingFaceFW/fineweb", split="train", streaming=True)

# Take first N documents (change as needed)
sample_size = 200000
subset = islice(dataset, sample_size)

doc_path = os.path.join("fineweb_data", "fineweb_en_2M_docs.txt")
sent_path = os.path.join("fineweb_data", "fineweb_en_sentences.txt")

# Step 1: Save document-level corpus
with open(doc_path, "w", encoding="utf-8") as f:
    for item in tqdm(subset, total=sample_size, desc="Saving documents"):
        text = item.get("text", "").replace("\n", " ").strip()
        if text:
            f.write(text + "\n")

# Step 2: Split into sentences and save
with open(doc_path, "r", encoding="utf-8") as fin, open(sent_path, "w", encoding="utf-8") as fout:
    for doc in tqdm(fin, desc="Splitting into sentences"):
        sentences = re.split(r'[.!?]\s+', doc.strip())
        for s in sentences:
            if len(s) > 20:
                fout.write(s.strip() + "\n")

print("✅ Saved datasets:")
print(f" - Documents: {doc_path}")
print(f" - Sentences: {sent_path}")
```
